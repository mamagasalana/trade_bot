```python
tmp =  a.get_raw('USDCHF', '1h', 90, 100, 1.0)
# profit1 = tmp.groupby(["direction"]).profit.sum().to_dict()
# profit1_count = tmp.groupby(["direction"]).profit.count().to_dict()

# group by consecutive direction streak
tmp["groupref"] = (tmp['direction'] != tmp['direction'].shift()).cumsum()
tmp2 = tmp.groupby(["groupref", "direction"], as_index=False).profit.sum()
# ret = tmp2.groupby('direction', group_keys=False, dropna=False).apply(a.score, include_groups=False)

# tmp2.direction.value_counts()
g= tmp2[tmp2.direction==-1]
p = g['profit']
wins = p[p > 0]
losses = p[p <= 0]
wr = wins.size / p.size if p.size else np.nan
avg_win = wins.mean() if wins.size else 0.0
avg_loss = -losses.mean() if losses.size else 0.0  # positive magnitude
sum_w = wins.sum()
sum_l = -losses.sum()  # positive magnitude
pf = (sum_w / sum_l) if sum_l > 0 else np.inf
payoff = (avg_win / avg_loss) if avg_loss > 0 else np.inf
pd.Series({
    'n': p.size,
    'win_rate': wr,
    'count_win' : wins.size,
    'count_loss' : losses.size,
    'sum_win' : sum_w,
    'sum_loss' : sum_l, 
    'avg_win': avg_win,
    'avg_loss': avg_loss,
    'payoff_ratio': payoff,
    'profit_factor': pf,
    'expectancy_per_trade': p.mean(),
    'median': p.median(),
    'iqr': p.quantile(0.75) - p.quantile(0.25),
    'std': p.std(ddof=1)
})
```

```python
id_cols = ['ccy', 'slow', 'fast']              # keep whatever identifiers you have
pf_cols = ['profit_factor_-1','profit_factor_0','profit_factor_1']

df_long = (
    df[id_cols + pf_cols]
      .melt(id_vars=id_cols, value_vars=pf_cols,
            var_name='pf_col', value_name='profit_factor')
      .dropna(subset=['profit_factor'])
)

# extract regime as int from the column name
df_long['regime'] = (
    df_long['pf_col'].str.extract(r'profit_factor_(-?1|0)')[0].astype(int)
)

# (optional) tidy up
df_long = df_long.drop(columns='pf_col')
df_long['log_pf'] = np.log(df_long['profit_factor'])

```

```python
from statsmodels.stats.proportion import proportion_confint

for reg in [-1,0,1]:
    x = df_long.loc[df_long.regime==reg, 'profit_factor'].dropna().values
    pct = (x > 1).mean()
    lo, hi = proportion_confint((x>1).sum(), len(x), method='wilson')
    print(f"Regime {reg}: median PF={np.median(x):.3f}, IQR=[{np.quantile(x,0.25):.3f},{np.quantile(x,0.75):.3f}], "
          f'%>1 = {100*pct:.1f}% (95% CI {100*lo:.1f}–{100*hi:.1f}%)')
```
