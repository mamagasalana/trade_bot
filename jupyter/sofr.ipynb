```python
%cd /home/ytee/test/trade_bot
import requests
import os
import json


FOLDER = 'files/newyorkfed'
os.makedirs(FOLDER, exist_ok=True)
FILE = os.path.join(FOLDER, 'rates.json')
if not os.path.exists(FILE):
    s = requests.get('https://markets.newyorkfed.org/api/rates/all/search.json?startDate=2011-01-01&endDate=2026-01-01&type=rate')
    with open(FILE, 'w') as ofile:
        json.dump(s.json(), ofile)

js =json.load(open(FILE, 'r'))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


df=  pd.DataFrame([x for x in js['refRates'] if  x['type'] =='SOFR'])
df.set_index('effectiveDate', inplace=True)

df2 = df[[x for x in df.columns if 'Percentile' in x]].replace('NA', np.nan).astype(float).sort_index().copy()

df2.plot()

_ =plt.xticks(rotation=45)
```
