```python
%cd /home/ytee/test/trade_bot/

from src.csv.macromicro import MM

app = MM()
df  = app.get_json("https://sc.macromicro.me/charts/35559/china-credit-impulse-index")

```

```python
import matplotlib.pyplot as plt

fig, ax1 = plt.subplots(figsize=(12, 6))
ax2 = ax1.twinx()
ax3 = ax1.twinx()
ax3.spines["right"].set_position(("outward", 60))

axes = [ax1, ax2, ax3]


for col, a , color in zip(df.columns, axes, ['red', 'green', 'blue']):
    df.dropna()[col].plot(ax=a, label=col, color=color)

lines = []
labels = []
for a in axes:
    lines += a.get_lines()
    labels += [line.get_label() for line in a.get_lines()]

ax1.legend(lines, labels, loc="best")
plt.show()
```
